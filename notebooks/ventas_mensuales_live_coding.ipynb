# Predicción de ventas mensuales utilizando un modelo SARIMA

```python
import pandas as pd
import numpy as np
import itertools
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.tsa.stattools import acf
from statsmodels.graphics.tsaplots import plot_pacf
from statsmodels.tsa.arima.model import ARIMA
```

## Exploración de los datos

```python
datapath = "../../src/data/"
```

```python
dataset = pd.read_csv(
    f"{datapath}stores_sales_forecasting.csv",
    encoding="latin-1",
    encoding_errors="replace"
)
dataset.head()
```

```python
dataset.describe(include='all').T
```

```python
dataset.dtypes
```

```python
dataset['Order Date'] = pd.to_datetime(dataset['Order Date'])
```

```python
dataset['Order Date'].unique()
```

```python
agg_sales = dataset.groupby(pd.Grouper(key='Order Date', freq='W'))['Sales'].sum()
```

```python
agg_sales.shape
```

```python
agg_sales.describe()
```

```python
agg_sales.head()
```

```python
agg_sales.plot(figsize=(10, 7), xlabel='Fecha', ylabel="Ventas semanales")
```
